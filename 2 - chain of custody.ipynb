## **Aim**
To write a python program for demonstrating Locard exchange principle

## **Algorithm**
**Step 1:** Start

**Step 2:** Define `get_hash(filename)` to read the file and compute its SHA-256 hash

**Step 3:** Define `timestamp()` to return the current time as `HH:MM:SS`

**Step 4:** Prompt the user to enter the file name and the number of custody transfers to record

**Step 5:** For each custody transfer, prompt for the custodian name and the action performed

**Step 6:** Compute the current SHA-256 hash of the file and record custodian, action, timestamp, and hash as a custody entry

**Step 7:** If this is not the first entry, compare the current hash with the previous entry's hash, printing "Integrity Verified" if they match or "Evidence Tampered" if they differ

**Step 8:** Wait for the user to confirm before moving to the next transfer, then once all transfers are recorded, print the full chain of custody log in order

**Step 9:** If the file is not found at any point, print "File not found." and End

```python
import hashlib
import time
from datetime import datetime


def get_hash(filename):
    with open(filename, "rb") as file:
        data = file.read()
    return hashlib.sha256(data).hexdigest()


def timestamp():
    return datetime.now().strftime("%H:%M:%S")


filename = input("Enter file name: ")

try:
    custody_log = []

    num_transfers = int(input("Enter number of custody transfers to record: "))

    for i in range(num_transfers):
        print(f"\n--- Custody Entry {i + 1} ---")
        custodian = input("Custodian name: ")
        action = input("Action (e.g. Collected, Transferred, Analyzed, Stored): ")

        current_hash = get_hash(filename)

        entry = {
            "custodian": custodian,
            "action": action,
            "timestamp": timestamp(),
            "hash": current_hash,
        }
        custody_log.append(entry)

        print(f"Recorded at {entry['timestamp']}")
        print(f"SHA-256 Hash: {entry['hash']}")

        if i > 0:
            previous_hash = custody_log[i - 1]["hash"]
            if current_hash == previous_hash:
                print("Integrity Verified: Hash unchanged since last entry")
            else:
                print("WARNING: Hash changed since last entry - Evidence Tampered")

        input("Press Enter once ready for the next custody step...")

    print("\n=== Full Chain of Custody ===")
    for i, entry in enumerate(custody_log, start=1):
        print(f"{i}. [{entry['timestamp']}] {entry['custodian']} - {entry['action']}")
        print(f"   Hash: {entry['hash']}")

except FileNotFoundError:
    print("File not found.")
```
