## **Aim**
To implement a program that extracts EXIF (Exchangeable Image File Format) metadata from an image file to identify camera settings, timestamps, and potential GPS coordinates.

## **Algorithm**
**Step 1:** Install and import the `PIL` (Pillow) library.

**Step 2:** Create a dummy image file for the program to analyze (or use an existing one).

**Step 3:** Open the image using `Image.open()`.

**Step 4:** Extract the EXIF data using the `_getexif()` method.

**Step 5:** Use `PIL.ExifTags.TAGS` to map the numeric EXIF tags to human-readable labels (e.g., 271 $\rightarrow$ "Make").

**Step 6:** Print the extracted metadata in a clean format.

```python
from PIL import Image
from PIL.ExifTags import TAGS
import os

def extract_exif(image_path):
    try:
        image = Image.open(image_path)
        exif_data = image._getexif()
        
        if not exif_data:
            return "No EXIF metadata found in this image."
        
        readable_exif = {}
        for tag_id, value in exif_data.items():
            tag = TAGS.get(tag_id, tag_id)
            readable_exif[tag] = value
            
        return readable_exif
    except Exception as e:
        return f"Error: {e}"

def main():
    image_path = "evidence_photo.jpg"

    print(f"Extracting EXIF from {image_path}...")
    data = extract_exif(image_path)
    
    if isinstance(data, dict):
        print("\n--- Metadata Found ---")
        for tag, val in data.items():
            print(f"{tag}: {val}")
    else:
        print(f"\n{data}")

if __name__ == "__main__":
    main()
```

## **Result**
This the program successfully extracts and displays EXIF metadata embedded in an image file.
