## **Aim**
To implement a program that calculates and compares MD5, SHA-1, and SHA-25 la6 hashes of a file to ensure data integrity and identify collisions or changes.

## **Algorithm**
**Step 1:** Import the `hashlib` library.

**Step 2:** Create a dummy evidence file.

**Step 3:** Implement a function that reads the file in binary mode and computes all three hashes (MD5, SHA-1, SHA-256) in a single pass.

**Step 4:** Display the calculated hashes.

**Step 5:** Modify the dummy file slightly (e.g., append one byte).

**Step 6:** Re-calculate the hashes and compare them with the original set to demonstrate how a single bit change affects all hash values.

```python
import hashlib
import os

def calculate_all_hashes(file_path):
    md5 = hashlib.md5()
    sha1 = hashlib.sha1()
    sha256 = hashlib.sha256()
    
    with open(file_path, "rb") as f:
        while chunk := f.read(4096):
            md5.update(chunk)
            sha1.update(chunk)
            sha256.update(chunk)
            
    return {
        "MD5": md5.hexdigest(),
        "SHA-1": sha1.hexdigest(),
        "SHA-256": sha256.hexdigest()
    }

def main():
    file_name = "hash_test.bin"
    
    # Create original file
    with open(file_name, "wb") as f:
        f.write(b"Forensic Evidence Data 2026")
    
    print(f"--- Original File: {file_name} ---")
    orig_hashes = calculate_all_hashes(file_name)
    for algo, val in orig_hashes.items():
        print(f"{algo}: {val}")
        
    # Modify file slightly (Slight corruption simulation)
    with open(file_name, "ab") as f:
        f.write(b"!") 
        
    print(f"\n--- Modified File (1 byte added) ---")
    mod_hashes = calculate_all_hashes(file_name)
    for algo, val in mod_hashes.items():
        print(f"{algo}: {val}")
        
    print("\n--- Comparison ---")
    for algo in ["MD5", "SHA-1", "SHA-256"]:
        if orig_hashes[algo] != mod_hashes[algo]:
            print(f"{algo} Hash changed! Integrity Compromised.")
        else:
            print(f"{algo} Hash matches.")

if __name__ == "__main__":
    main()
```

## **Result**
This the program successfully calculates and compares MD5, SHA-1, and SHA-256 hash values for a digital evidence file.
