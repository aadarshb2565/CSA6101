## **Aim**
To implement a program that extracts file metadata including file name, size, creation time, modification time, and access time.

## **Algorithm**
**Step 1:** Import `os`, `datetime`, and `stat` libraries.

**Step 2:** Define a function `get_file_metadata(filepath)` that uses `os.stat()` to retrieve file status.

**Step 3:** Extract the following attributes from the stat result:
   - File name (from path)
   - File size in bytes (st_size)
   - Creation time (st_ctime on Windows, st_birthtime on Unix if available)
   - Modification time (st_mtime)
   - Access time (st_atime)
   - File permissions/mode (st_mode)
   - Owner UID/GID (st_uid, st_gid)

**Step 4:** Convert timestamp values to human-readable format using `datetime.fromtimestamp()`.

**Step 5:** Format and display the metadata in a structured table.

**Step 6:** In `main()`, test with multiple files of different types.

```python
import os
import stat
from datetime import datetime

def get_file_metadata(filepath):
    try:
        st = os.stat(filepath)
        
        metadata = {
            "file_name": os.path.basename(filepath),
            "full_path": os.path.abspath(filepath),
            "size_bytes": st.st_size,
            "size_human": format_size(st.st_size),
            "creation_time": datetime.fromtimestamp(st.st_ctime).strftime("%Y-%m-%d %H:%M:%S"),
            "modification_time": datetime.fromtimestamp(st.st_mtime).strftime("%Y-%m-%d %H:%M:%S"),
            "access_time": datetime.fromtimestamp(st.st_atime).strftime("%Y-%m-%d %H:%M:%S"),
            "permissions": stat.filemode(st.st_mode),
            "owner_uid": st.st_uid,
            "owner_gid": st.st_gid,
            "inode": st.st_ino,
            "links": st.st_nlink,
            "is_directory": stat.S_ISDIR(st.st_mode),
            "is_file": stat.S_ISREG(st.st_mode),
            "is_symlink": stat.S_ISLNK(st.st_mode)
        }
        return metadata
    except Exception as e:
        return {"error": str(e)}

def format_size(bytes_val):
    for unit in ['B', 'KB', 'MB', 'GB']:
        if bytes_val < 1024:
            return f"{bytes_val:.2f} {unit}"
        bytes_val /= 1024
    return f"{bytes_val:.2f} TB"

def display_metadata(meta):
    if "error" in meta:
        print(f"Error: {meta['error']}")
        return
    
    print(f"\n{'='*50}")
    print(f"File Metadata: {meta['file_name']}")
    print(f"{'='*50}")
    for key, value in meta.items():
        label = key.replace('_', ' ').title()
        print(f"  {label:<20}: {value}")

def main():
    test_files = [
        "test.txt",
        "1 - locard.ipynb",
        "evidence_photo.jpg"
    ]
    
    # Create a test file
    with open("test.txt", "w") as f:
        f.write("Test file for metadata extraction")
    
    for file in test_files:
        if os.path.exists(file):
            meta = get_file_metadata(file)
            display_metadata(meta)
        else:
            print(f"\nFile not found: {file}")

if __name__ == "__main__":
    main()
```

## **Result**
This the program successfully extracts file metadata including file name, size, creation time, modification time, and access time.
