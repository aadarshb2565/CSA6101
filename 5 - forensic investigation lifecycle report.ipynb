## **Aim**
To simulate the documentation process of a digital forensics investigation by tracking the lifecycle phases: Identification, Preservation, Analysis, and Documentation.

## **Algorithm**
**Step 1:** Define a `ForensicCase` class to store case details and phase logs.

**Step 2:** Implement methods to log activity for each phase (Identification, Preservation, Analysis, Documentation).

**Step 3:** Include a timestamp for every action to maintain a proper audit trail (Chain of Custody).

**Step 4:** Create a dummy case scenario involving a compromised workstation.

**Step 5:** Progress the case through the four lifecycle phases, adding specific findings at each step.

**Step 6:** Generate a final formatted report summary.

```python
from datetime import datetime

class ForensicCase:
    def __init__(self, case_id, investigator):
        self.case_id = case_id
        self.investigator = investigator
        self.logs = {
            "Identification": [],
            "Preservation": [],
            "Analysis": [],
            "Documentation": []
        }

    def add_log(self, phase, detail):
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        self.logs[phase].append(f"[{timestamp}] {detail}")

    def generate_report(self):
        report = f"FORENSIC INVESTIGATION REPORT\nCase ID: {self.case_id}\nInvestigator: {self.investigator}\n"
        report += "="*30 + "\n"
        for phase, entries in self.logs.items():
            report += f"\n## {phase}\n"
            if not entries:
                report += "No activities recorded.\n"
            else:
                for entry in entries:
                    report += f"- {entry}\n"
        return report

def main():
    # Initialize Case
    case = ForensicCase("CASE-2026-001", "Agent Smith")
    
    # Phase 1: Identification
    case.add_log("Identification", "Identified compromised laptop (Dell XPS 13, S/N: 12345).")
    case.add_log("Identification", "User reported unauthorized file modifications.")
    
    # Phase 2: Preservation
    case.add_log("Preservation", "Created bit-stream image of the hard drive.")
    case.add_log("Preservation", "Calculated SHA-256 hash of the original disk: e3b0c442...")
    
    # Phase 3: Analysis
    case.add_log("Analysis", "Found hidden directory /home/user/.hidden_logs.")
    case.add_log("Analysis", "Recovered deleted bash history showing ssh connection to external IP 192.168.1.50.")
    
    # Phase 4: Documentation
    case.add_log("Documentation", "Compiled findings into a final legal report.")
    case.add_log("Documentation", "Verified hash integrity of the evidence image.")
    
    print(case.generate_report())

if __name__ == "__main__":
    main()
```

## **Result**
This the program successfully generates a structured digital forensic investigation lifecycle report.
