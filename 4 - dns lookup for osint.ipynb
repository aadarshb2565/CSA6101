## **Aim**
To implement a program that performs DNS lookups for a given domain to gather Open Source Intelligence (OSINT) regarding its infrastructure.

## **Algorithm**
**Step 1:** Import the `socket` library for network operations.

**Step 2:** Define the target domain to be analyzed.

**Step 3:** Use `socket.gethostbyname()` to resolve the domain to its IPv4 address.

**Step 4:** Use `socket.gethostbyaddr()` to find the hostname associated with the IP.

**Step 5:** Handle potential errors (like `socket.gaierror` or `socket.herror`) using try-except blocks to ensure the program doesn't crash on invalid domains.

**Step 6:** Display the collected IP and Hostname information.

```python
import socket

def perform_dns_lookup(domain):
    results = {"Domain": domain, "IP Address": "Not found", "Hostname": "Not found"}
    
    try:
        # Get IP from Domain
        ip_address = socket.gethostbyname(domain)
        results["IP Address"] = ip_address
        
        # Get Hostname from IP
        hostname = socket.gethostbyaddr(ip_address)[0]
        results["Hostname"] = hostname
        
    except socket.gaierror:
        print(f"Error: Could not resolve domain {domain}")
    except socket.herror:
        print(f"Error: Could not resolve hostname for IP")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        
    return results

def main():
    # Test domains
    targets = ["google.com", "example.com"]
    
    for target in targets:
        print(f"\n--- OSINT DNS Lookup for {target} ---")
        info = perform_dns_lookup(target)
        for key, value in info.items():
            print(f"{key}: {value}")

if __name__ == "__main__":
    main()
```

## **Result**
This the program successfully performs DNS lookups for OSINT reconnaissance.
