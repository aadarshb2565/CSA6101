## **Aim**
To implement a Python program that detects potential phishing emails by analyzing content for common phishing keywords and suspicious patterns.

## **Algorithm**
**Step 1:** Define a list of suspicious keywords commonly found in phishing emails (e.g., "urgent", "password", "verify", "bank", "suspended").

**Step 2:** Create a dummy email content string for testing.

**Step 3:** Convert the email content to lowercase to ensure case-insensitive matching.

**Step 4:** Iterate through the keyword list and check if any keyword exists in the email content.

**Step 5:** Calculate a "phishing score" based on the number of matches.

**Step 6:** If the score exceeds a predefined threshold, flag the email as "Suspicious"; otherwise, mark it as "Safe".

**Step 7:** Print the analysis results.

```python
import re

def detect_phishing(text):
    # Common phishing keywords
    keywords = ["urgent", "password", "verify", "bank", "suspended", "account", "login", "update", "action required"]
    
    text_lower = text.lower()
    matches = [word for word in keywords if word in text_lower]
    
    score = len(matches)
    threshold = 2
    
    is_phishing = score >= threshold
    return is_phishing, matches, score

def main():
    # Dummy email content
    email_content = """
    Subject: URGENT: Your Bank Account has been Suspended!
    Dear User, 
    We have detected unusual activity on your account. 
    Please verify your password immediately to avoid permanent suspension.
    Click here to login: http://secure-bank-verify.com/login
    """
    
    print("--- Analyzing Email Content ---")
    print(f"Content:\n{email_content}")
    
    is_phishing, matches, score = detect_phishing(email_content)
    
    print("\n--- Analysis Result ---")
    print(f"Phishing Score: {score}")
    print(f"Keywords Found: {', '.join(matches)}")
    print(f"Verdict: {'SUSPICIOUS' if is_phishing else 'SAFE'}")

if __name__ == "__main__":
    main()
```

## **Result**
This the program successfully detects phishing emails.
