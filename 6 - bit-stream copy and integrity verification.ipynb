## **Aim**
To implement a program that simulates a bit-stream image copy of a file and verifies its integrity using SHA-256 hashing.

## **Algorithm**
**Step 1:** Create a dummy "original" evidence file with random content.

**Step 2:** Implement a function to calculate the SHA-256 hash of a file.

**Step 3:** Perform a bit-stream copy (binary read/write) of the original file to a new "image" file.

**Step 4:** Calculate the hash of the original file and the copied image file.

**Step 5:** Compare the two hashes.

**Step 6:** If hashes match, the integrity is verified; otherwise, it's flagged as corrupted.

```python
import hashlib
import os

def calculate_sha256(file_path):
    sha256_hash = hashlib.sha256()
    with open(file_path, "rb") as f:
        # Read in chunks to handle large files
        for byte_block in iter(lambda: f.read(4096), b""):
            sha256_hash.update(byte_block)
    return sha256_hash.hexdigest()

def create_bitstream_copy(source, destination):
    with open(source, "rb") as src:
        with open(destination, "wb") as dst:
            dst.write(src.read())

def main():
    original_file = "evidence_disk.bin"
    image_file = "evidence_image.bin"
    
    # Create dummy evidence file
    with open(original_file, "wb") as f:
        f.write(os.urandom(1024)) # 1KB of random data
    
    print(f"Created original evidence file: {original_file}")
    
    # Step 1: Original Hash
    orig_hash = calculate_sha256(original_file)
    print(f"Original SHA-256: {orig_hash}")
    
    # Step 2: Perform Bit-stream Copy
    create_bitstream_copy(original_file, image_file)
    print(f"Bit-stream copy created: {image_file}")
    
    # Step 3: Image Hash
    image_hash = calculate_sha256(image_file)
    print(f"Image SHA-256:    {image_hash}")
    
    # Verification
    if orig_hash == image_hash:
        print("\nINTEGRITY VERIFIED: Hashes match. The copy is a perfect duplicate.")
    else:
        print("\nINTEGRITY FAILED: Hashes do not match!")

if __name__ == "__main__":
    main()
```

## **Result**
This the program successfully creates a bit-stream forensic copy and verifies its integrity.
