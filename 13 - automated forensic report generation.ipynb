## **Aim**
To implement a program that aggregates findings from various forensic analysis steps (hashes, logs, file types) and automatically generates a structured final forensic report.

## **Algorithm**
**Step 1:** Define a `Finding` data structure to hold the evidence ID, category, and description.

**Step 2:** Create a list of simulated findings gathered from previous "notebooks" (e.g., a phishing email, a port scan, a recovered file).

**Step 3:** Implement a `ReportGenerator` class that takes these findings.

**Step 4:** Format the report with a header (Case ID, Date, Investigator).

**Step 5:** Categorize findings into sections (e.g., "Network Analysis", "File System Analysis").

**Step 6:** Export the final report to a text file.

```python
from datetime import datetime

class Finding:
    def __init__(self, category, evidence_id, description):
        self.category = category
        self.evidence_id = evidence_id
        self.description = description

class ReportGenerator:
    def __init__(self, case_id, investigator):
        self.case_id = case_id
        self.investigator = investigator
        self.findings = []

    def add_finding(self, finding):
        self.findings.append(finding)

    def generate_txt_report(self, filename):
        with open(filename, "w") as f:
            f.write("="*50 + "\n")
            f.write(f"AUTOMATED DIGITAL FORENSICS REPORT\n")
            f.write(f"Case ID: {self.case_id}\n")
            f.write(f"Investigator: {self.investigator}\n")
            f.write(f"Date: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
            f.write("="*50 + "\n\n")
            
            categories = set(f.category for f in self.findings)
            for cat in sorted(categories):
                f.write(f"SECTION: {cat}\n")
                f.write("-" * 20 + "\n")
                for find in self.findings:
                    if find.category == cat:
                        f.write(f"ID: {find.evidence_id} | {find.description}\n")
                f.write("\n")
            
            f.write("\n--- END OF REPORT ---\n")

def main():
    report = ReportGenerator("CASE-2026-99", "Lead Analyst")
    
    # Simulate findings from other notebooks
    report.add_finding(Finding("Network", "NET-01", "Port scanning detected from 10.0.0. la0.20 (10 unique ports)."))
    report.add_finding(Finding("Email", "EML-05", "Phishing email identified with keywords: urgent, suspended."))
    report.add_finding(Finding("Disk", "DSK-12", "Recovered deleted JPEG file at offset 500."))
    report.add_finding(Finding("Disk", "DSK-13", "Verified SHA-256 integrity of evidence image."))
    
    output_file = "final_forensic_report.txt"
    report.generate_txt_report(output_file)
    
    print(f"Report successfully generated: {output_file}")
    with open(output_file, "r") as f:
        print("\n--- Report Preview ---\n")
        print(f.read())

if __name__ == "__main__":
    main()
```

## **Result**
This the program successfully generates a structured digital forensic report from investigation findings.
